```python
import cv2
import time

cap = cv2.VideoCapture(0) 

start_time = time.time()  

while True:
    ret, frame = cap.read()  
    if ret:
        an_time = time.time() - start_time 
        
        if an_time < 5:  # İlk 5 saniye boyunca renkli görüntü göster
            cv2.imshow('Webcam', frame)
            frame_copy = frame.copy() 
            font=cv2.FONT_HERSHEY_SIMPLEX
            text="Renkli Mod"
            cv2.putText(frame_copy, text, (10, 30), font, 1, (128, 0, 128), 2, cv2.LINE_AA)
            cv2.imshow('Webcam', frame_copy)

        else:
            # 5 saniye sonra siyah-beyaz görüntüye geç
            gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
            cv2.imshow('Webcam', gray)
            gray_copy=gray.copy()
           
            font = cv2.FONT_HERSHEY_SIMPLEX
            text = "Siyah Beyaz Mod"
            cv2.putText(gray_copy, text, (10, 30), font, 1, (0,0, 255), 2, cv2.LINE_AA)
            cv2.imshow('Webcam',gray_copy)
    
    # 'q' tuşuna basınca çık
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release() 
cv2.destroyAllWindows()  

```
