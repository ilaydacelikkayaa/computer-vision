```python
import cv2
import time

# Haar Cascade yüz tespit sınıflandırıcıyı yükleyelim
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

# Webcam açılır
cap = cv2.VideoCapture(0)

# Başlangıç zamanını al
start_time = time.time()

while True:
    ret, frame = cap.read() 
    if not ret:
        break  
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    
   
    elapsed_time = time.time() - start_time

    if elapsed_time > 3:  
        faces = face_cascade.detectMultiScale(gray, 1.1, 4)
        for (x, y, w, h) in faces:
            face = frame[y:y+h, x:x+w] #x ve y sol üst w,h genişlik ve yükseklik
            
        
            gray_copy = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
            frame = cv2.cvtColor(gray_copy, cv2.COLOR_GRAY2BGR)  # Tüm görüntüyü siyah-beyaz yap

            frame[y:y+h, x:x+w] = face 

    else:
        # 3 saniye geçmeden önce sadece siyah-beyaz görüntü göster
        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        frame = cv2.cvtColor(frame, cv2.COLOR_GRAY2BGR)

    # Görüntüyü göster
    cv2.imshow('Webcam', frame)

    # 'q' tuşuna basınca çık
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()  # Kamera bağlantısını kapat
cv2.destroyAllWindows()  # Pencereyi kapat
```
